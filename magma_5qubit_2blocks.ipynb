# 5-qubit perfect code [[5,1,3]]

$$
S = <XZZXI, IXZZX, XIXZZ, ZXIXZ>
$$

```python
n = 5
k = 1
d = 3

stabs = ['XZZXI','IXZZX','XIXZZ','ZXIXZ']
H_symp = stabs_to_H_symp(stabs)
```

```python
command_file_name    = f"magma_commands_n{n}k{k}d{d}.txt"
output_file_name     = f"magma_output_n{n}k{k}d{d}.txt"
with open(output_file_name, "w") as output_file:
        pass

# magma_obj = qec_code_auts_from_magma_with_intersection(n, k, d, H_symp)
magma_obj = qec_code_auts_from_magma(n, k, d, H_symp)
magma_obj.run(fileroot='./',save_magma_commands=True)
run_magma_online_from_file(command_file_name)
code_auts_dict = parse_magma_output(output_file_name, magma_obj)
pprint.pprint(code_auts_dict)
print("generators = ", len(code_auts_dict["auts"]))
```

```python
auts = code_auts_dict['auts']
circuits = []
symp_mats = []
for num, aut in enumerate(auts):
    phys_act        = circ_from_aut(H_symp,aut)        
    phys_circ, _    = phys_act.circ()
    log_act         = logical_circ_and_pauli_correct(H_symp,phys_circ)
    circ            = log_act.run()
    circuits.append(circ)
    symp_mats.append(log_act.U_logical_act())

for i, circ in enumerate(circuits):
    logical_act, circ_desc = circ
    print(f"Circuit ({i})", "Logical Action:", logical_act)
    # display(construct_circuit(circ_desc, n).draw(output='mpl'))
```

## Automorphisms of the two-block code

For the 10-qubit code we use two copies of the 5-qubit code

$C_{10}=C_5\oplus C_5$.

Each 5-qubit block has automorphism group size

$|Aut(C_5)|=360$.

Since the two blocks are independent, we can apply any automorphism to each block separately:

$Aut(C_5)\times Aut(C_5)$

which gives

$360\cdot360=129600$.

Because the two blocks are identical, we can also **swap the blocks**, giving an extra factor of $2$:

$|Aut(C_{10})| = 360^2\cdot 2 = 259200$.

So the automorphism group is

$Aut(C_{10}) \cong (Aut(C_5)\times Aut(C_5))\rtimes S_2$  

or equivalently

$A_6 \wr S_2$.

### Generator interpretation

Using the 3-bit representation:

block 1 → columns $1\!-\!15$  
block 2 → columns $16\!-\!30$

Generators fall into three types:

- act only on block 1 → logical gates on qubit 1  
- act only on block 2 → logical gates on qubit 2  
- mix the blocks → logical SWAP

### Logical gates obtained

Automorphisms produce logical operations such as

$(H,S,X\sqrt{})_1,\quad (H,S,X\sqrt{})_2,\quad SWAP$.

These correspond to

$(\text{Clifford}_1 \otimes \text{Clifford}_2)$

and possibly

$SWAP$.

The direct-sum code therefore does **not naturally produce entangling logical gates** like CNOT or CZ.
